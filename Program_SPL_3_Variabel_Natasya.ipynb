```python
import numpy as np

# Membuat matriks koefisien
A = np.array([
    [2, 1, -1],
    [-3, -1, 2],
    [-2, 1, 2]
], dtype=float)

# Membuat matriks konstanta
B = np.array([8, -11, -3], dtype=float)

# Menyelesaikan SPL menggunakan metode eliminasi Gauss
def eliminasi_gauss(A, B):
    n = len(B)

    # Membuat matriks gabungan
    M = np.column_stack((A, B))

    # Proses eliminasi
    for i in range(n):
        # Pivot parsial
        pivot = i + np.argmax(np.abs(M[i:, i]))
        M[[i, pivot]] = M[[pivot, i]]

        # Mengubah elemen di bawah pivot menjadi nol
        for j in range(i + 1, n):
            faktor = M[j, i] / M[i, i]
            M[j, i:] = M[j, i:] - faktor * M[i, i:]

    # Proses substitusi mundur
    x = np.zeros(n)

    for i in range(n - 1, -1, -1):
        x[i] = (M[i, n] - np.dot(M[i, i+1:n], x[i+1:n])) / M[i, i]

    return x

# Menampilkan hasil
hasil = eliminasi_gauss(A, B)

print("Hasil Penyelesaian SPL 3 Variabel:")
print("Nilai x =", hasil[0])
print("Nilai y =", hasil[1])
print("Nilai z =", hasil[2])
```
