````python
import numpy as np

# Membuat matriks koefisien
A = np.array([
    [2, 1, -1],
    [-3, -1, 2],
    [-2, 1, 2]
], dtype=float)

# Membuat matriks konstanta
B = np.array([8, -11, -3], dtype=float)

# Menyelesaikan SPL menggunakan metode eliminasi Gauss
def eliminasi_gauss(A, B):
    n = len(B)

    # Membuat matriks gabungan
    M = np.column_stack((A, B))

    # Proses eliminasi
    for i in range(n):
        # Pivot parsial
        pivot = i + np.argmax(np.abs(M[i:, i]))
        M[[i, pivot]] = M[[pivot, i]]

        # Mengubah elemen di bawah pivot menjadi nol
        for j in range(i + 1, n):
            faktor = M[j, i] / M[i, i]
            M[j, i:] = M[j, i:] - faktor * M[i, i:]

    # Proses substitusi mundur
    x = np.zeros(n)

    for i in range(n - 1, -1, -1):
        x[i] = (M[i, n] - np.dot(M[i, i+1:n], x[i+1:n])) / M[i, i]

    return x

# Menampilkan hasil
hasil = eliminasi_gauss(A, B)

print("Hasil Penyelesaian SPL 3 Variabel:")
print("Nilai x =", hasil[0])
print("Nilai y =", hasil[1])
print("Nilai z =", hasil[2])
````

6.1 menentukan Pivot Prasial

**Kode Program:**

```python
pivot = i + np.argmax(np.abs(M[i:, i]))
M[[i, pivot]] = M[[pivot, i]]
```

**Penjelasan:**

Bagian ini digunakan untuk menentukan elemen pivot dengan metode Partial Pivoting.

Fungsi `np.abs()` digunakan untuk mendapatkan nilai absolut dari elemen matriks. Selanjutnya, `np.argmax()` digunakan untuk mencari posisi elemen yang memiliki nilai absolut terbesar.

Nilai tersebut digunakan untuk menentukan baris pivot. Kemudian, perintah `M[[i, pivot]] = M[[pivot, i]]` digunakan untuk menukar posisi baris agar elemen pivot terbesar berada pada posisi yang sedang diproses.

Tujuannya adalah membantu menjaga kestabilan perhitungan numerik.

6.2 mengubah Elemen Di Bawah Pivot Menjadi Nol

**Kode Program:**

```python
for j in range(i + 1, n):
    faktor = M[j, i] / M[i, i]
    M[j, i:] = M[j, i:] - faktor * M[i, i:]
```

**Penjelasan:**

Bagian ini digunakan untuk melakukan proses eliminasi maju atau Forward Elimination.

Perintah `for j in range(i + 1, n):` digunakan untuk melakukan perulangan pada baris yang berada di bawah baris pivot.

Variabel `faktor` digunakan untuk menghitung nilai pengali dalam proses eliminasi. Perhitungan dilakukan dengan membagi elemen pada posisi baris j dan kolom i dengan elemen pivot.

Selanjutnya, perintah `M[j, i:] = M[j, i:] - faktor * M[i, i:]` digunakan untuk mengurangi baris yang sedang diproses dengan hasil perkalian faktor dan baris pivot.

Proses tersebut bertujuan mengubah elemen di bawah diagonal utama menjadi nol sehingga membentuk matriks segitiga atas.
